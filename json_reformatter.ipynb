```python
import os
import nbformat
from nbformat import NotebookNode

# Get current working directory
PATH = os.getcwd()

# Iterate through all .ipynb files
for filename in os.listdir(PATH):
    if filename.endswith(".ipynb"):
        file_path = os.path.join(PATH, filename)
        print(f"Processing: {filename}")

        try:
            # Load the notebook
            nb = nbformat.read(file_path, as_version=4)  # you can use v5 too

            # Check integrity (nbformat will throw error if invalid)
            # Remove widgets metadata if it exists
            if "widgets" in nb.metadata:
                del nb.metadata["widgets"]
                print(f"  Removed metadata.widgets from {filename}")

            # Optional: remove widgets from each cell too
            for cell in nb.cells:
                if "widgets" in cell.metadata:
                    del cell.metadata["widgets"]
                    print(f"  Removed metadata.widgets from a cell in {filename}")

            # Save the cleaned notebook
            nbformat.write(nb, file_path)
            print(f"  Saved cleaned notebook: {filename}")

        except Exception as e:
            print(f"  ERROR loading {filename}: {e}")

print("Done cleaning all notebooks in folder.")
```
